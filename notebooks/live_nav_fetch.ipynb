```python
import os
import requests
import pandas as pd
```

```python
data = {}
urls = ["https://api.mfapi.in/mf/125497","https://api.mfapi.in/mf/119551","https://api.mfapi.in/mf/120503","https://api.mfapi.in/mf/118632","https://api.mfapi.in/mf/119092","https://api.mfapi.in/mf/120841"]
```

```python
def fetch_data(url):
    response = requests.get(url,timeout=30)
    response.raise_for_status()
    data = response.json()
    return data
fetch_data(urls[0])
```

```python
def call_number_of_urls(urls):
    data = {}
    for i in range(len(urls)):
        stored_data = fetch_data(urls[i])
        data[f'{stored_data["meta"]["fund_house"]}'] =stored_data
    return data
```

```python
all_df = {}
def convert_data(urls):
    all_data = call_number_of_urls(urls)
    data_keys = all_data.keys()
    for i in data_keys:
        all_df[i] = pd.DataFrame(all_data[i]["data"])
        all_df[i]["date"] = pd.to_datetime(all_df[i]["date"],dayfirst=True)
        all_df[i]["nav"] = pd.to_numeric(all_df[i]["nav"],errors="coerce")
    return all_df
```

```python
check_null_dict = {}
def check_null(urls):
    all_data = convert_data(urls)
    all_data_keys = all_data.keys()
    for keys in all_data_keys:
        check_null_dict[keys] = all_data[keys].isnull().sum()
    return check_null_dict
```

```python
check_null(urls)
```

```python
def save_data(urls):
    all_data = convert_data(urls)
    all_data_keys = all_data.keys()
    for keys in all_data_keys:
        
        file_path = os.path.join("C:/Users/ratna/Documents/mutual_funds_analysis/data/raw",f"{keys}.csv")
        all_data[keys].to_csv(file_path,index=False)
        print(f"{keys}.csv is  SAVED SUCCESSFULLY")
        
```

```python
save_data(urls)
```
