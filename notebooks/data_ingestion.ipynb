```python
def print_shape():
    for i in data_keys:
        print(f"Shape of {i} = ",data[i].shape)
```

```python
def print_dtypes():
    for i in data_keys:
        print("="*50)
        print("Data Type Of =",i)
        print("-"*50)
        print(data[i].dtypes)
        print("="*50,"\n")
```

```python
def print_heads():
    for i in data_keys:
        print("="*140)
        print("Data Type Of =",i)
        print("-"*140)
        print("\n",data[i].head())
        print("="*140,"\n")
```

```python
def any_null():
    for i in data_keys:
        print("="*140)
        print(i.upper()," = ")
        print(data[i].isnull().sum())
        print("="*140)
any_null()
```

```python
def any_duplicates():
    for i in data_keys:
        print("="*140)
        print(i,"'s duplicated values = ",data[i].duplicated().sum())
        print("="*140)
any_duplicates()
```
